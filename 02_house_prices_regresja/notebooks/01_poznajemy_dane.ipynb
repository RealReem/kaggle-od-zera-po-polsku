### 🔍 Rozbiórka kodu na czynniki pierwsze (Składnia Pythona dla laika)

Kiedy uruchomisz powyższą komórkę (`Shift + Enter`), zobaczysz wymiary obu tabel oraz fragment wielkiej tabeli z domami. Rozszyfrujmy każdy znak, który napisaliśmy:

* **`# ...` (Kratka / Hasz)** — wszystko po tym znaku w danej linijce to **komentarz**. Komputer całkowicie go ignoruje. Piszesz go tylko dla siebie jako notatkę.
* **`import pandas as pd`** — czysty Python sam w sobie jest jak pusty warsztat. Jeśli chcemy pracować z tabelami (jak w Excelu), musimy przynieść specjalną skrzynkę z narzędziami o nazwie **`pandas`**. Słowo `as pd` oznacza: *„od teraz nazywaj tę skrzynkę w skrócie `pd`, bo nie chce mi się za każdym razem pisać pełnej nazwy `pandas`”*.
* **`train = pd.read_csv('../data/train.csv')`** — tutaj dzieje się magia:
  * **Kropka `.`** w `pd.read_csv` oznacza **„sięgnij do środka”**. Mówimy: *„Otwórz skrzynkę `pd` i wyjmij z niej narzędzie o nazwie `read_csv` (czytaj plik CSV)”*.
  * **Nawiasy okrągłe `()`** w Pythonie oznaczają **AKCJĘ (wykonaj czynność / uruchom funkcję)**. Do środka nawiasu wkładamy w cudzysłowie ścieżkę `'../data/train.csv'` (`..` oznacza wyjście poziom wyżej z folderu `notebooks/`, a następnie wejście do folderu `data/`).
  * **Znak `=`** w programowaniu NIE oznacza „równa się” jak na matematyce! Oznacza **„zapisz wynik z prawej strony do pudełka po lewej stronie”**. Tworzymy pudełko (zmienną) o nazwie `train` i wkładamy do niego wczytaną tabelę.
* **`train.shape` (dlaczego tu NIE MA nawiasów okrągłych?)** — `shape` (kształt/wymiar) to nie jest *czynność*, którą trzeba wykonać, tylko **cecha (właściwość)** tabeli, jak kolor samochodu czy wzrost człowieka. Dlatego po kropce piszemy samo `.shape` bez `()`.
  * Wynik `(1460, 81)` oznacza: **1460 wierszy (domów)** i **81 kolumn (cech)**. 
  * Zauważ, że `test.shape` daje `(1459, 80)` — ma o **1 kolumnę mniej** (80 zamiast 81). Której kolumny brakuje? Oczywiście ceny `SalePrice`, którą mamy odgadnąć!
* **`train.head()` (dlaczego tu SĄ nawiasy okrągłe?)** — `head` (głowa/początek) to polecenie: *„utnij i pokaż mi pierwsze 5 wierszy tabeli”*. Ponieważ to czynność (metoda), na końcu stawiamy `()`. Jeśli wpiszesz w nawias np. `train.head(10)`, pokaże 10 pierwszych wierszy.

```python
# Wyciągamy narzędzie do rysowania wykresów (matplotlib)
import matplotlib.pyplot as plt

# 1. Podsumowanie statystyczne kolumny z ceną (SalePrice)
print(train['SalePrice'].describe())

# 2. Rysujemy histogram, żeby zobaczyć rozkład cen własnymi oczami
plt.figure(figsize=(10, 5))
plt.hist(train['SalePrice'], bins=40, color='skyblue', edgecolor='black')

# Dodajemy pionowe linie dla średniej (czerwona) i mediany (zielona)
srednia = train['SalePrice'].mean()
mediana = train['SalePrice'].median()
plt.axvline(srednia, color='red', linestyle='--', linewidth=2, label=f'Średnia: {srednia:,.0f} $')
plt.axvline(mediana, color='green', linestyle='-', linewidth=2, label=f'Mediana: {mediana:,.0f} $')

# Podpisujemy wykres, żeby był czytelny
plt.title('Rozkład cen domów (SalePrice) w zbiorze treningowym')
plt.xlabel('Cena sprzedaży w dolarach ($)')
plt.ylabel('Liczba domów')
plt.legend()
plt.show()
```

### 📊 Nawiasy kwadratowe vs okrągłe oraz zagadka: Średnia kontra Mediana

#### 1. Dlaczego piszemy `train['SalePrice']` w nawiasach KWADRATOWYCH `[]`?
To złota zasada Pythona, którą warto zapamiętać od pierwszego dnia:
* **Nawiasy okrągłe `()`** = **„Zrób coś!”** (uruchom funkcję, np. `.describe()`, `.head()`, `print()`).
* **Nawiasy kwadratowe `[]`** = **„Wyciągnij coś ze środka!”** (wybierz kolumnę lub wyfiltruj wiersze). Pisząc `train['SalePrice']`, mówimy: *„Z całej wielkiej tabeli `train` wyciągnij mi tylko jedną kolumnę o nazwie `'SalePrice'`”*.

---

#### 2. Co mówi nam `.describe()` i wykres?
Metoda `.describe()` (opisz) błyskawicznie liczy dla nas najważniejsze liczby:
* **`count` (1460)** — dla ilu domów mamy podaną cenę (dla wszystkich 1460, nie ma pustych miejsc!).
* **`min` (34 900 $)** — najtańszy dom w bazie.
* **`max` (755 000 $)** — najdroższa rezydencja w bazie.
* **`mean` (180 921 $)** — **średnia arytmetyczna** (suma cen wszystkich domów podzielona przez 1460).
* **`50%` (163 000 $)** — **mediana** (wartość środkowa).

#### 💡 Co oznacza to, że ŚREDNIA (180 921 \$) jest wyraźnie większa od MEDIANY (163 000 \$)?
Wyobraź sobie, że w małym barze siedzi 10 zwykłych osób i każda zarabia 5 000 zł. 
* Jaka jest **mediana** (zarobki osoby stojącej dokładnie pośrodku)? **5 000 zł**.
* Jaka jest **średnia**? Też **5 000 zł**.

Nagle do baru wchodzi miliarder. 
* **Mediana** (osoba w samym środku kolejki od najbiedniejszego do najbogatszego) **nadal wynosi 5 000 zł** — opisuje typowego gościa w barze.
* Ale **średnia** nagle skacze do **kilkuset milionów złotych**, bo jeden miliarder „pociągnął” średnią w górę!

Spójrz na nasz wykres (histogram): większość domów kosztuje w okolicach **130 000 – 200 000 \$** (tam słupki są najwyższe), ale po prawej stronie ciągnie się tzw. **długi prawy ogon** — pojedyncze luksusowe rezydencje za 500 000 \$ czy 755 000 \$. To właśnie te nieliczne wille „ciągną” czerwoną linię (średnią) w prawo względem zielonej linii (mediany).
